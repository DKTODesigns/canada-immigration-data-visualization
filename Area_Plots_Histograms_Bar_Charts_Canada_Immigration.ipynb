# Area Plots, Histograms, and Bar Charts

In this lab, I will use the Canada immigration dataset to create different types of visualizations:

- Area plots
- Histograms
- Bar charts

The dataset shows immigration to Canada by country from 1980 to 2013.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## Load the Dataset

I uploaded the Canada immigration CSV file into Google Colab.  
Now I will load it into a Pandas DataFrame.

```python
# Load the CSV file
df = pd.read_csv("Canada.csv")

# Display the first five rows
df.head()
```

```python
# Check the shape of the dataset
df.shape
```

```python
# Check the column names
df.columns
```

## Set Country as the Index

Setting the Country column as the index makes it easier to select rows by country name.

```python
# Set Country as the index
df.set_index("Country", inplace=True)

# Display the first few rows to confirm the change
df.head()
```

```python
# Create a list of years from 1980 to 2013
years = list(map(str, range(1980, 2014)))

years
```

## Area Plot: Top 5 Countries

An area plot shows trends over time.  
Here, I will visualize the top 5 countries that contributed the most immigrants to Canada from 1980 to 2013.

```python
# Sort the dataset by Total immigration values
df.sort_values(["Total"], ascending=False, axis=0, inplace=True)

# Select the top 5 countries
df_top5 = df.head(5)

# Transpose the dataframe so years are on the x-axis
df_top5 = df_top5[years].transpose()

df_top5.head()
```

```python
# Create an area plot
df_top5.plot(kind="area", alpha=0.35, figsize=(12, 6))

plt.title("Immigration Trend of Top 5 Countries")
plt.ylabel("Number of Immigrants")
plt.xlabel("Years")

plt.show()
```

## Area Plot: Bottom 5 Countries

Now I will create an area plot for the 5 countries with the lowest total immigration to Canada.

```python
# Select the bottom 5 countries
df_bottom5 = df.tail(5)

# Transpose the dataframe
df_bottom5 = df_bottom5[years].transpose()

df_bottom5.head()
```

```python
# Create an area plot for bottom 5 countries
df_bottom5.plot(kind="area", alpha=0.45, figsize=(12, 6))

plt.title("Immigration Trend of Bottom 5 Countries")
plt.ylabel("Number of Immigrants")
plt.xlabel("Years")

plt.show()
```

## Histogram: Immigration Distribution in 2013

A histogram shows the frequency distribution of numerical data.  
Here, I will look at how immigration values were distributed across countries in 2013.

```python
# Create a histogram for immigration in 2013
df["2013"].plot(kind="hist", figsize=(10, 6))

plt.title("Histogram of Immigration from 195 Countries in 2013")
plt.ylabel("Number of Countries")
plt.xlabel("Number of Immigrants")

plt.show()
```

```python
# Get the bin counts and bin edges
count, bin_edges = np.histogram(df["2013"])

print(count)
print(bin_edges)
```

```python
# Create a histogram with custom x-axis ticks
df["2013"].plot(kind="hist", figsize=(10, 6), xticks=bin_edges)

plt.title("Histogram of Immigration from 195 Countries in 2013")
plt.ylabel("Number of Countries")
plt.xlabel("Number of Immigrants")

plt.show()
```

## Histogram: Denmark, Norway, and Sweden

Now I will compare the immigration distribution for Denmark, Norway, and Sweden from 1980 to 2013.

```python
# Select Denmark, Norway, and Sweden
df_dns = df.loc[["Denmark", "Norway", "Sweden"], years]

# Display the data
df_dns
```

```python
# Transpose the dataframe
df_dns = df_dns.transpose()

df_dns.head()
```

```python
# Create an overlapping histogram
df_dns.plot(kind="hist", figsize=(10, 6), alpha=0.6, bins=15)

plt.title("Histogram of Immigration from Denmark, Norway, and Sweden")
plt.ylabel("Number of Years")
plt.xlabel("Number of Immigrants")

plt.show()
```

## Bar Chart: Iceland Immigration

A bar chart is useful for comparing values across categories.  
Here, I will visualize immigration from Iceland to Canada from 1980 to 2013.

```python
# Select Iceland data
df_iceland = df.loc["Iceland", years]

df_iceland.head()
```

```python
# Create a vertical bar chart
df_iceland.plot(kind="bar", figsize=(12, 6))

plt.title("Icelandic Immigrants to Canada from 1980 to 2013")
plt.xlabel("Year")
plt.ylabel("Number of Immigrants")

plt.show()
```

## Bar Chart with Annotation

The bar chart shows an increase around the time of the 2008–2011 financial crisis.  
I will add an annotation to highlight this period.

```python
# Create the bar chart
df_iceland.plot(kind="bar", figsize=(12, 6))

plt.title("Icelandic Immigrants to Canada from 1980 to 2013")
plt.xlabel("Year")
plt.ylabel("Number of Immigrants")

# Add annotation
plt.annotate(
    "2008–2011 Financial Crisis",
    xy=(30, 70),
    xytext=(20, 120),
    arrowprops=dict(arrowstyle="->")
)

plt.show()
```

## Horizontal Bar Chart: Top 15 Countries

A horizontal bar chart is useful when category labels are long.  
Here, I will show the top 15 countries with the highest total immigration to Canada.

```python
# Select top 15 countries
df_top15 = df["Total"].head(15)

df_top15
```

```python
# Create horizontal bar chart
df_top15.plot(kind="barh", figsize=(10, 8))

plt.title("Top 15 Countries Contributing to Immigration to Canada")
plt.xlabel("Number of Immigrants")
plt.ylabel("Country")

plt.show()
```
